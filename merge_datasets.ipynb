```python
import pandas as pd

# Load datasets
df1 = pd.read_csv("tweets1.csv")
df2 = pd.read_excel("cleaned_reddit_comments.csv.xlsx")

# Tweets dataset (labels: "Hate"/"Non-Hate")
df1 = df1.rename(columns={"text": "text", "label": "label"})
df1["label"] = (
    df1["label"].astype(str).str.strip().str.lower().map({"non-hate": 0, "hate": 1})
)
df1 = df1.dropna(subset=["label"])
df1["label"] = df1["label"].astype(int)

# Reddit dataset (labels: "non hate"/"hate")
df2 = df2.rename(columns={"translated_text": "text", "Unnamed: 1": "label"})
df2["label"] = (
    df2["label"].astype(str).str.strip().str.lower().map({"non hate": 0, "hate": 1})
)
df2 = df2.dropna(subset=["label"])
df2["label"] = df2["label"].astype(int)

# Merge both datasets
merged_df = pd.concat([df1, df2], ignore_index=True)

# Drop missing values
merged_df = merged_df.dropna()

# Save merged dataset
merged_df.to_csv("merged_dataset.csv", index=False)

print("✅ Merged dataset saved as merged_dataset.csv")
print(merged_df.head())
print("Shape:", merged_df.shape)
print("Label dtype:", merged_df['label'].dtype)
```
